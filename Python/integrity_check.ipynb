```python
import os 
import filecmp 
import re

# Function to generalize subject IDs in file paths
def generalize_subject_ids(file_path):
    # Replace patterns like sub-001, sub-10, etc., with <SUBJECT>
    generalized_path = re.sub(r'sub-\d+', '<SUBJECT>', file_path)
    return generalized_path

# Recursively get all relative file paths in both folders
def get_all_files(folder):
    file_list = []
    if not os.path.exists(folder):
        #print(f"ERROR: Folder does not exist: {folder}")
        return file_list

    #print(f"Scanning folder: {folder}")
    for root, dirs, files in os.walk(folder):
        #print(f"Currently in: {root}")
        #print(f"Subdirectories: {dirs}")
        #print(f"Files: {files}")

        for file in files:
            # Get relative path to maintain structure
            rel_path = os.path.relpath(os.path.join(root, file), folder)
            # Generalize subject IDs in the relative path
            generalized_path = generalize_subject_ids(rel_path)
            #print(f"Found file: {generalized_path}")
            file_list.append(generalized_path)

    # If still empty, let the user know
    if not file_list:
        print(f"No files found in folder: {folder}")

    return file_list

# Compare the generalized file lists
def compare_folders(template_folder, test_folder):
    template_files = get_all_files(template_folder)
    test_files = get_all_files(test_folder)

    print("\n=== Comparison Results ===")

    # Compare file presence
    missing_in_test = set(template_files) - set(test_files)
    missing_in_template = set(test_files) - set(template_files)

    if missing_in_test:
        print("\nFiles present in Template but missing in Test:")
        for file in missing_in_test:
            print(file)
    else:
        print("\nNo files missing in Test folder!")

    if missing_in_template:
        print("\nFiles present in Test but missing in Template:")
        for file in missing_in_template:
            print(file)
    else:
        print("\nNo files missing in Template folder!")

    # Compare file sizes for common files
    common_files = set(template_files) & set(test_files)
    print("\n=== File Size Comparison for Common Files ===")
    for file in common_files:
        template_file_path = os.path.join(template_folder, file.replace("<SUBJECT>", "sub-XXX"))  # Placeholder for real ID
        test_file_path = os.path.join(test_folder, file.replace("<SUBJECT>", "sub-XXX"))  # Placeholder for real ID
        
        # Check if the files exist in their original paths
        if os.path.exists(template_file_path) and os.path.exists(test_file_path):
            template_size = os.path.getsize(template_file_path)
            test_size = os.path.getsize(test_file_path)
            if template_size != test_size:
                print(f"Size mismatch: {file}")
                print(f"  Template size: {template_size} bytes")
                print(f"  Test size: {test_size} bytes")
        else:
            print(f"Warning: Could not find files for size comparison: {file}")


# Set your folder paths
template_folder = os.path.abspath("/vols/Scratch/pfr545/Amb_project/prf_test_H/derivatives/fmriprep/sub-002")
test_folder = os.path.abspath("/vols/Scratch/pfr545/Amb_project/prf_test/derivatives/fmriprep/sub-10")

# Run the comparison
compare_folders(template_folder, test_folder)
```
