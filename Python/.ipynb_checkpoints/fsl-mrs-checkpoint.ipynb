```python
## then we create functions that can be used in a loop 

### function to check the files 
def check_twix_file(folder_path, extension): # 
    for file_name in os.listdir(folder_path):
        print("Processing file:", file_name)
        if file_name.endswith(extension):
            !spec2nii twix -v -q $file_name

### function to convert the files 
def convert_twix_file(folder_path, extension, pattern='*', loc ='V1', recursive_search=False): # 
    # Get the full path of the source folder
    folder_path = os.path.abspath(folder_path) + os.sep
    for file_name in os.listdir(folder_path):
        print("Processing file:", file_name)
        if file_name.endswith(extension):
            if re.search(loc, file_name):
                new_file1 = 'steam_metab_raw_V1'
                new_file2 = 'steam_ecc_raw_V1'
                new_file3 = 'steam_wref_comb_raw_V1'
                new_file4 = 'steam_wref_quant_raw_V1'
            elif re.search('M1', file_name):
                new_file1 = 'steam_metab_raw_M1'
                new_file2 = 'steam_ecc_raw_M1'
                new_file3 = 'steam_wref_comb_raw_M1'
                new_file4 = 'steam_wref_quant_raw_M1'
            else:
                new_file1 = 'steam_metab_raw'
                new_file2 = 'steam_ecc_raw'
                new_file3 = 'steam_wref_comb_raw'
                new_file4 = 'steam_wref_quant_raw'
            if re.search(pattern, file_name):
                !spec2nii twix -e image $file_name -f $new_file1 -o $folder_path
                !spec2nii twix -e phasecor $file_name -f $new_file2 -o $folder_path 
            elif re.search('wref1', file_name):
                !spec2nii twix -e image $file_name -f $new_file3 -o $folder_path
            elif re.search('wref3', file_name):
                !spec2nii twix -e image $file_name -f $new_file4 -o $folder_path
                
                


### function to convert the files 
def glance_file(folder_path, extension): # 
    for file_name in os.listdir(folder_path):
        if file_name.endswith(extension):
            new_file = os.path.splitext(file_name)[0]+'.png'
            print("Processing file:", file_name) 
            !mrs_tools vis $file_name --save $new_file
            !mrs_tools info $file_name 


### function to move specific files (more flexible this way)
def move_files(folder_path, folder_destination, pattern='*', recursive_search=False):
    # Create the destination folder if it doesn't exist
    os.makedirs(folder_destination, exist_ok=True)
    # Get the full path of the source folder
    folder_path = os.path.abspath(folder_path) + os.sep
    # Get the full path of the destination folder
    folder_destination = os.path.abspath(folder_destination) + os.sep
    # Find files matching the pattern in the source folder
    files = glob.glob(os.path.join(folder_path, '**' if recursive_search else '', pattern), recursive=recursive_search)
    # Move each file to the destination folder
    for file_path in files:
        relative_path = os.path.relpath(file_path, folder_path)
        destination_path = os.path.join(folder_destination, relative_path)
        os.makedirs(os.path.dirname(destination_path), exist_ok=True)
        shutil.move(file_path, destination_path)




## main function from Betina
```
